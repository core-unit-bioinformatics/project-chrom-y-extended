```python
import pathlib as pl
import pandas as pd

# that is a dump from results table -> sheet "1_Samples"
hg_annot_file = pl.Path(
    "/home/ebertp/work/code/cubi/project-chrom-y-extended/annotation/raw",
    "samples_hg.csv"
)

hg_output_file = pl.Path(
    "/home/ebertp/work/code/cubi/project-chrom-y-extended/annotation/norm",
    "samples_hg.tsv"
)

column_idx = [0, 3, 4, 7, 8, 9]
column_name = ["sample", "population", "continental_group", "cohort", "hg_short", "hg_long"]

rows = []
with open(hg_annot_file) as listing:
    for ln, line in enumerate(listing, start=1):
        if ln < 4:
            continue
        if not line.strip():
            continue
        if ln > 145:
            break
        columns = line.strip().split(",")
        if columns[0] == "HG005":
            columns[3] = "CHS"
        row = []
        for i, n in zip(column_idx, column_name):
            val = columns[i].strip()
            if val == "CEPH_pedigree":
                val = "ceph"
            if val == "HGSVC3":
                val = "hgsvc"
            if val == "HPRC" or val == "T2T?" or val == "T2T":
                val = "hprc"
            row.append(val)
        rows.append(tuple(row))

df = pd.DataFrame.from_records(rows, columns=column_name)
df.sort_values("sample", inplace=True)
df.to_csv(hg_output_file, sep="\t", header=True, index=False)
```
